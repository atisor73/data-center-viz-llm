| Column | Meaning |
|---|---|
| facility_name | Name or label of the data center/project. |
| address | Street address, road, parcel description, or approximate site location. |
| city | City or locality associated with the facility. |
| state | Two-letter U.S. state abbreviation. |
| zip | ZIP code. Treat as text, not a number. |
| county | County where the facility is located. |
| lat | Latitude coordinate used for map plotting. |
| long | Longitude coordinate used for map plotting. |
| status | Development/operation status: Operating, Proposed, Approved/Permitted/Under construction, Expanding, Suspended, or Cancelled. |
| location_confidence | Confidence in the mapped location: mostly High, Medium, or Low. |
| purpose | Stated or inferred use, such as AI, cloud computing, Bitcoin/crypto, telecom, or mixed use. Sparse field. |
| operator_name | Company/operator/developer associated with the facility. |
| tenant | Known tenant or end customer, when different from operator. Very sparse. |
| mw | Power capacity in megawatts, when known. Values can be exact or ranges/text, so parse carefully. |
| sizerank | Size category derived from MW, e.g. small, medium, large, hyperscale, mega campus. |
| power_source | Known or reported energy source, e.g. grid, natural gas, solar, nuclear, hybrid. Sparse. |
| dedicated_power_plant | Notes whether the project has or may have a dedicated power plant or named power provider. |
| number_of_generators | Backup or onsite generator count/details. Often free text, not clean numeric data. |
| number_of_buildings | Number of buildings planned or present, when known. |
| cooling_source | Cooling resource type, e.g. water, air, or hybrid. Sparse. |
| cooling_type | Cooling system type, e.g. closed loop, open loop, fans. Sparse. |
| facility_size_sqft | Building/facility square footage. Often numeric with commas. |
| property_size_acres | Land area in acres. |
| project_cost | Reported project investment/cost, usually text like $1 billion. |
| expected_date_online | Expected opening/operation date, usually year or year range. |
| community_pushback | Indicates known public/community opposition. In this file, populated values are Yes; blanks should mean unknown/not documented, not necessarily “No.” |
| advocacy_information | Narrative notes about local advocacy, opposition, public hearings, moratoria, lawsuits, etc. |
| resistance_status | Categorical community response level, currently mainly Organized Advocacy or Emerging Concern. |
| nda | Notes about nondisclosure agreements or secrecy concerns tied to the project. Very sparse. |
| community_group_website_1 | First URL or label for a related community group. |
| community_group_website_2 | Second community group URL/label. |
| petition_url | Petition, campaign, or action URL related to the facility/project. |
| other_info | Miscellaneous notes that do not fit other fields. Often important free-text context. |
| information_source | Broad source category: Media Monitoring, PEC, Sci4GA, Crowdsourced, or Other. |
| info_source_1 | Primary source URL/document. |
| info_source_2 | Additional source URL/document. |
| info_source_3 | Additional source URL/document. |
| info_source_4 | Additional source URL/document. |
| info_source_5 | Additional source URL/document. |
| info_source_6 | Additional source URL/document. |
| info_source_7 | Additional source URL/document. |
| info_source_8 | Additional source URL/document. |
| date_created | Date the row was added to the dataset. |
| date_updated | Date the row was last updated. |
